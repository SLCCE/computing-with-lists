# 2D Lists

Sometimes our data is not one-dimensional. For example, a tic-tac-toe board, which has 3 rows and 3 columns. So, to store it a normal list is inconvenient. A solution to this issue is a 2D list.

```python
a = [[1, 2, 3],
     [4, 5, 6],
     [7, 8, 9]]
```

This 2D list has 3 rows and 3 columns.

## Indexing
You can think of a 2D list as a list of lists. So your first index will get you your first list. By indexing into this list, we can get an individual element.

```python
a = [[1, 2, 3],
     [4, 5, 6],
     [7, 8, 9]]
firstRow = a[0]
firstRowSecondElement = a[0][1]
# prints 2
print(firstRowSecondElement)
```

Like with the 1D list, we can both index into it and modify values.

```python
a = [[1, 2, 3],
     [4, 5, 6],
     [7, 8, 9]]
# change 2 to 10
a[0][1] = 10
print(a)
```
