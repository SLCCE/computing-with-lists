# Operations
There are many things we can do with a list that make it very useful.

## Adding to a List
insert(index, object): inserts object at index

```python
a = [2, 4, 6, 8, 10]
a.insert(0, -2)
print(a)
```

append(object): adds an object at the end of a list

## Removing from a List
remove(value): removes the first instance of a value

```python
a = [2, 4, 6, 8, 10]
a.remove(2)
print(a)
```

note: if we try to remove something not in the list there will be an error

```python
a = [2, 4, 6, 8, 10]
a.remove(0)
print(a)
```

pop(index): removes the value at an index and returns it

```python
a = [2, 4, 6, 8, 10]
x = a.pop(0)
print(a, x)
# we also don't necessarily need to use the popped value
# default pop is the last index
a.pop()
print(a)
```

## Sorting a List

```python
a = [1, 4, 2, 8, 5, 7]
print(a)
a.sort()
print(a)
```

## Finding the Minimum and Maximum

```python
a = [1, 4, 2, 8, 5, 7]
mn = min(a)
mx = max(a)
print("The minimum is", mn)
print("The maximum is", mx)
```

Example: sort characters in a word

```python
s = "antigravitational"
letters = list(s)
letters.sort()
print(letters)
```

## Slicing
*(consider deleting)*

Slicing is for when we want a section of the list.
start:stop:step (optional)

The stop is non-inclusive

Example: First 3 elements

```python
a = [1, 4, 2, 8, 5, 7]
x = a[:3]
print(x)
```

Example: reverse a list

```python
a = [1, 4, 2, 8, 5, 7]
a = a[::-1]
```
