# Introduction to Lists

A list is a ordered data structure that holds values at indexes.

For example, integers

```python
a = [2, 4, 6, 8, 10]
```

strings

```python
a = ["1", "2", "3", "4", "5"]
```

both

```python
a = [1, 2, "3", 4, 5]
```

An index is like a page number on a book where the first page is page 0.
We can access items in a list by indexing into it.

| index | value |
| -- | -- |
| 0 | 2 |
| 1 | 4 |
| 2 | 6 |
| 3 | 8 |
| 4 | 10 |

```python
a = [2, 4, 6, 8, 10]
print(a[0])
print(a[1])
print(a[2])
print(a[3])
print(a[4])
```
